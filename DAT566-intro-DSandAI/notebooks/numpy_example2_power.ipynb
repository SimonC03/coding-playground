# NumPy Example 2: Power

The following matrix represents the electricity consumption of 8 different buildings (rows) across a week (columns Monday to Sunday).

We want to compute averages of electricity consumption and identify inefficient buildings.

```python
import numpy as np

# rows = buildings
# columns = days (Mon Tue Wed Thu Fri Sat Sun)
usage = np.array([
    [28, 30, 29, 31, 32, 30, 29],  # Building 0
    [20, 22, 21, 23, 24, 18, 17],  
    [35, 36, 34, 37, 38, 36, 37],  
    [25, 26, 24, 27, 28, 20, 19],  
    [40, 42, 41, 43, 44, 42, 43],  
    [18, 19, 20, 21, 22, 16, 15],  
    [30, 31, 32, 33, 34, 31, 32],  
    [22, 23, 24, 25, 26, 24, 23],  # Building 7
])
```

## Average usage
- Compute weekly averages per building

```python
avg_usage = usage.mean(1)
avg_usage
```

## Electricity usage during weekdays and weekends
- Select power usage during weekdays (Mon-Fri)
- Select power usage during weekends (Sat-Sun)
- Compute respective averages per building

```python
usage_weekdays = usage[:,:5]
usage_weekdays
```

```python
usage_weekends = usage[:,5:]
usage_weekends
```

```python
avg_usage_weekdays = usage_weekdays.mean(1)
avg_usage_weekdays
```

```python
avg_usage_weekends = usage_weekends.mean(1)
avg_usage_weekends
```

## Find inefficient buildings
A building is considered inefficient if
- the daily average is > 30 kWh **and**
- the usage does not go down during weekends

```python
inefficient = (avg_usage > 30) & (avg_usage_weekends >= avg_usage_weekdays)
inefficient
```

## Weekly usage of inefficient buildings
- List daily usage for inefficient buildings

```python
daily_inefficient = usage[inefficient,:]
daily_inefficient
```

## Most daily usage
- Which day has the most total usage over all days?
- List the usage for that day for all buildings

```python
daily_total = usage.sum(0)
daily_total
```

```python
max_usage_day = np.argmax(daily_total)
max_usage_day
```

```python
usage_max_day = usage[max_usage_day,:]
usage_max_day
```
